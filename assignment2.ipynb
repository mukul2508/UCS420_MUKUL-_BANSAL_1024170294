```python

import random
random.seed(1024170294)
rand_list = [random.randint(100, 900) for _ in range(1, 100)]
print(rand_list)
```

```python
print([i for i in rand_list if i%2!=0])
print(len([i for i in rand_list if i%2!=0]))
```

```python
print([i for i in rand_list if i%2==0])
print(len([i for i in rand_list if i%2==0]))
```

```python
def isprime(n):
  for i in range(2,n):
    if n%i==0:
            return False
    return True

primes=[i for i in rand_list if isprime(i)]
print(primes);
```

```python
count_list = [rand_list.count(i) for i in rand_list]
max_count = max(count_list)
for i in range(len(rand_list)):
    if count_list[i] == max_count and rand_list[i] not in rand_list[:i]:
        print(rand_list[i], "occurs", max_count, "times")
```
